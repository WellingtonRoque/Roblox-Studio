## 2. O cenário da aula: a missão do explorador

O explorador chegou a uma região da Ilha do Tesouro. Para avançar, ele precisa:

1. Começar com uma quantidade de moedas, chaves, cristais e energia;
2. percorrer uma trilha com etapas;
3. encontrar recursos durante a caminhada;
4. verificar se possui uma chave para abrir o portão;
5. consultar o inventário;
6. receber uma mensagem de conclusão.

Nesta aula, os eventos serão simulados pelo código. Ainda não vamos usar interface gráfica, detecção de toque, comunicação entre Scripts ou sistemas de salvamento.

## 3. Organizando os dados com variáveis e tabelas

Vamos criar uma tabela chamada `explorador`. Ela armazenará os dados principais do personagem.

```python
local explorador = {
    nome = "Aventureiro",
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100
}

print("Nome:", explorador.nome)
print("Moedas:", explorador.moedas)
print("Chaves:", explorador.chaves)
print("Cristais:", explorador.cristais)
print("Energia:", explorador.energia)
```

A tabela reúne informações relacionadas ao mesmo personagem. Para acessar cada valor, utilizamos o nome do campo, por exemplo `explorador.moedas`.

## 4. Criando uma função para mostrar o status

Vamos criar uma função para exibir os dados do explorador sempre que necessário.

```python
local explorador = {
    nome = "Aventureiro",
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100
}

local function mostrarStatus()
    print("====== STATUS DO EXPLORADOR ======")
    print("Nome:", explorador.nome)
    print("Moedas:", explorador.moedas)
    print("Chaves:", explorador.chaves)
    print("Cristais:", explorador.cristais)
    print("Energia:", explorador.energia)
end

mostrarStatus()
```

A função `mostrarStatus()` concentra os comandos de exibição. Poderemos chamá-la no início, durante e no final da missão.

## 5. Tomando decisões com `if` e `else`

O explorador encontrou um portão. Para abri-lo, precisa ter pelo menos uma chave.

Vamos utilizar uma condicional para verificar essa situação.

```python
local explorador = {
    chaves = 1
}

if explorador.chaves > 0 then
    print("O portao foi aberto!")
else
    print("Voce precisa encontrar uma chave.")
end
```

A condição `explorador.chaves > 0` verifica se a quantidade de chaves é maior que zero.

- Se for verdadeira, o primeiro bloco será executado;
- caso contrário, será executado o bloco do `else`.

Experimente alterar o valor de `chaves` para `0` e executar novamente.

## 6. Criando uma função para coletar recursos

Vamos criar uma função que recebe a quantidade de moedas encontradas e atualiza o inventário.

```python
local explorador = {
    moedas = 10
}

local function coletarMoedas(quantidade)
    explorador.moedas = explorador.moedas + quantidade
    print("Voce encontrou", quantidade, "moedas.")
end

coletarMoedas(5)
coletarMoedas(3)

print("Total de moedas:", explorador.moedas)
```

A função utiliza um parâmetro para receber a quantidade. Assim, podemos reutilizar a mesma função para diferentes coletas.

## 7. Repetindo etapas com `for`

A trilha possui cinco etapas. Vamos utilizar `for` para representar o avanço do explorador.

```python
for etapa = 1, 5 do
    print("Explorando etapa:", etapa)
end
```

O laço começa em 1 e termina em 5. A variável `etapa` recebe a posição atual a cada repetição.

## 8. Percorrendo uma lista de eventos

Podemos guardar os nomes das etapas em uma tabela e percorrer a lista com `for`.

```python
local etapas = {
    "Floresta",
    "Rio",
    "Caverna",
    "Ponte",
    "Portao do Tesouro"
}

for i = 1, #etapas do
    print("Etapa", i, "-", etapas[i])
end
```

Aqui integramos dois conceitos:
- a tabela `etapas` armazena os nomes;
- o laço `for` percorre as posições da lista.

O operador `#etapas` informa quantos elementos existem.

## 9. Criando uma função para verificar o portão

Agora vamos combinar função, parâmetro, condicional e tabela. A função verifica se o explorador tem uma chave. Se tiver, consome uma chave e informa que o portão foi aberto.

```python
local explorador = {
    chaves = 1
}

local function abrirPortao()
    if explorador.chaves > 0 then
        explorador.chaves = explorador.chaves - 1
        print("Portao aberto!")
        return true
    else
        print("Voce nao tem chaves suficientes.")
        return false
    end
end

local conseguiuAbrir = abrirPortao()
print("Conseguiu abrir?", conseguiuAbrir)
```

A função retorna:
- `true` quando o portão é aberto;
- `false` quando não há chave disponível.

A variável `conseguiuAbrir` recebe o resultado devolvido por `return`.

## 10. Utilizando `while` para acompanhar a energia

Vamos simular uma caminhada em que o explorador perde energia a cada etapa. O laço continua enquanto houver energia maior que zero e enquanto não tiver ultrapassado cinco etapas.

```python
local energia = 30
local etapa = 1

while energia > 0 and etapa <= 5 do
    print("Etapa:", etapa)
    energia = energia - 10
    print("Energia restante:", energia)

    etapa = etapa + 1
end

print("Fim da caminhada.")
```

A condição combina duas verificações:
- `energia > 0`: o explorador ainda tem energia;
- `etapa <= 5`: ainda existem etapas a percorrer.

O operador `and` exige que as duas condições sejam verdadeiras para continuar o laço.

O contador `etapa` aumenta em cada repetição. Sem essa atualização, o laço poderia não terminar quando a energia não mudasse.

## 11. Projeto integrado — Missão na Ilha do Tesouro

Agora vamos reunir os conceitos em um único Script.

**Preparação no Roblox Studio:**
1. Abra o projeto Ilha do Tesouro;
2. no Explorer, abra `ServerScriptService`;
3. crie um `Script` chamado `MissaoIlha`;
4. copie o código completo abaixo;
5. clique em **Play**;
6. acompanhe a sequência no painel **Output**.

```python
-- =========================================
-- DADOS DO EXPLORADOR
-- =========================================

local explorador = {
    nome = "Aventureiro",
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100
}

-- =========================================
-- ETAPAS DA AVENTURA
-- =========================================

local etapas = {
    "Floresta",
    "Rio",
    "Caverna",
    "Ponte",
    "Portao do Tesouro"
}

-- =========================================
-- FUNCOES
-- =========================================

local function mostrarStatus()
    print("========== STATUS ==========")
    print("Explorador:", explorador.nome)
    print("Moedas:", explorador.moedas)
    print("Chaves:", explorador.chaves)
    print("Cristais:", explorador.cristais)
    print("Energia:", explorador.energia)
end

local function coletarMoedas(quantidade)
    explorador.moedas = explorador.moedas + quantidade
    print("Moedas encontradas:", quantidade)
end

local function coletarCristais(quantidade)
    explorador.cristais = explorador.cristais + quantidade
    print("Cristais encontrados:", quantidade)
end

local function abrirPortao()
    if explorador.chaves > 0 then
        explorador.chaves = explorador.chaves - 1
        print("O portao foi aberto!")
        return true
    else
        print("Falta uma chave para abrir o portao.")
        return false
    end
end

-- =========================================
-- INICIO DA MISSAO
-- =========================================

print("===== INICIO DA MISSAO =====")
mostrarStatus()

-- O explorador percorre as etapas
for i = 1, #etapas do
    print("----------------------------")
    print("Etapa", i, "-", etapas[i])

    -- Cada etapa consome 10 pontos de energia
    explorador.energia = explorador.energia - 10

    -- Eventos simulados durante a aventura
    if i == 2 then
        coletarMoedas(5)
    elseif i == 3 then
        coletarCristais(2)
    elseif i == 4 then
        coletarMoedas(10)
    end

    print("Energia:", explorador.energia)
end

-- Tentativa de abrir o portao
print("----------------------------")
print("Tentando abrir o portao...")

local portaoAberto = abrirPortao()

if portaoAberto then
    print("MISSAO CONCLUIDA!")
else
    print("MISSAO NAO CONCLUIDA.")
end

-- Resultado final
mostrarStatus()
print("===== FIM DA MISSAO =====")
```

### Entendendo o projeto integrado

Neste único Script, utilizamos:

| Conceito | Onde aparece |
|---|---|
| Variáveis | Dados e contadores do explorador |
| Tabelas | `explorador` e `etapas` |
| Funções | `mostrarStatus`, `coletarMoedas`, `coletarCristais`, `abrirPortao` |
| Parâmetros | `quantidade` nas funções de coleta |
| `return` | Resultado da tentativa de abrir o portão |
| `if / elseif / else` | Eventos e decisão de abrir o portão |
| `for` | Percorrer as etapas da aventura |
| `#` | Obter a quantidade de etapas |

**Resultado esperado:** o explorador termina as cinco etapas com 85 de energia, 25 moedas, 0 chaves e 2 cristais. Como começou com uma chave, consegue abrir o portão e concluir a missão.
