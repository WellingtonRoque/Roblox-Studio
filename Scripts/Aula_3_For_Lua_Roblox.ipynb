# 1. 🤔 Por que repetir?

Imagine escrever:

```lua
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
```

Funciona, mas seria trabalhoso fazer isso 100 vezes.

O `for` permite pedir ao computador para repetir uma ação.

# 2. 🔁 Primeiro `for`

A estrutura básica:

```lua
for i = 1, 5 do
    print("Moeda encontrada!")
end
```

- `for` inicia a repetição;
- `i` é o contador;
- `1` é o início;
- `5` é o fim;
- `do` inicia o bloco;
- `end` encerra o bloco.

Tudo entre `do` e `end` será repetido.

````lua
for i = 1, 5 do
    print("Moeda encontrada!")
end
````

# 3. 🔢 Conhecendo o contador `i`

```lua
for i = 1, 5 do
    print(i)
end
```

Resultado:

```text
1
2
3
4
5
```

O valor de `i` muda a cada repetição.

````lua
for i = 1, 5 do
    print(i)
end
````

# 4. 🪙 Contando moedas

```lua
for i = 1, 10 do
    print("Moeda número:", i)
end
```

O contador pode representar o número de cada objeto, etapa ou item do jogo.

````lua
for i = 1, 10 do
    print("Moeda número:", i)
end
````

# 6. ⏪ Contagem regressiva

Para diminuir o contador:


````lua
for i = 10, 1, -1 do
    print(i)
end
````

# 7. ➕ Pulando números

Podemos definir um passo:


````lua
for i = 2, 20, 2 do
    print(i)
end
````

# 9. 🧮 O contador em cálculos

O `i` pode participar de cálculos:

```lua

    local pontos = i * 10
```

Ainda não é um sistema de pontuação. Estamos apenas praticando o contador.

````lua
for i = 1, 5 do
    local pontos = i * 10
    print("Moeda:", i, "Pontos:", pontos)
end
````
