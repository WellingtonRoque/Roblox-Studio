## 2. Conhecendo os painéis Explorer e Properties

No Roblox Studio, o painel **Explorer** mostra a hierarquia de objetos do jogo. O painel **Properties** apresenta as propriedades do objeto selecionado.

Se esses painéis não estiverem visíveis, abra a guia **View** e habilite `Explorer` e `Properties`.

No Explorer, os objetos aparecem organizados em uma estrutura hierárquica. Por exemplo:

```text
Workspace
    Baseplate
    Part
ServerScriptService
    Script
ReplicatedStorage
StarterGui
StarterPlayer
```

Cada serviço tem uma finalidade. Nesta aula, vamos conhecer os locais mais importantes para organizar os Scripts.

## 3. Principais locais do Roblox Studio

### Workspace
Contém os objetos que fazem parte do mundo 3D: terreno, peças, modelos, portas, baús e outros elementos visíveis ou posicionados no cenário.

### ServerScriptService
Local recomendado para Scripts de servidor. É apropriado para regras importantes do jogo, como atualizar o inventário, controlar recompensas e validar a abertura de portões.

### ReplicatedStorage
Local para objetos e módulos que precisam estar disponíveis tanto para o servidor quanto para os clientes. Nesta aula, vamos apenas conhecer esse serviço; a comunicação entre servidor e cliente será estudada depois.

### StarterGui
Local em que ficam as interfaces que serão disponibilizadas ao jogador, como telas, indicadores e menus.

### StarterPlayer
Contém configurações e locais relacionados ao jogador. Dentro dele existem estruturas como `StarterPlayerScripts` e `StarterCharacterScripts`.

### StarterPack
Local utilizado para disponibilizar ferramentas ao jogador quando ele entra no jogo.

## 4. Diferença entre Script e LocalScript

O Roblox possui diferentes tipos de scripts. Vamos começar distinguindo os dois mais comuns nesta etapa.

| Tipo | Execução | Uso introdutório |
|---|---|---|
| `Script` | Servidor, conforme o local e a configuração | Regras do jogo, inventário, recompensas |
| `LocalScript` | Cliente de cada jogador, em locais permitidos | Interface e controles locais |

**Regra prática para nosso projeto:**
- Atualização de moedas, chaves e cristais: manter no servidor;
- regras de abertura de portão: manter no servidor;
- elementos visuais e controles da interface: poderão utilizar LocalScripts quando estudarmos interfaces.

Um `LocalScript` não executa em qualquer lugar. Ele precisa estar em um local permitido para execução no cliente. Por isso, nesta aula, os exemplos funcionais principais serão feitos com `Script` em `ServerScriptService`.

## 6. Script 1 — Dados do jogador

Na pasta `Jogador`, crie um `Script` chamado `DadosJogador`.

Neste primeiro exemplo, vamos guardar e mostrar os dados iniciais do explorador.

````python
local jogador = {
    nome = "Aventureiro",
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100
}

print("DADOS DO JOGADOR")
print("Nome:", jogador.nome)
print("Moedas:", jogador.moedas)
print("Chaves:", jogador.chaves)
print("Cristais:", jogador.cristais)
print("Energia:", jogador.energia)
````

Esse Script demonstra como organizar um arquivo responsável pelos dados do jogador.

**Observação importante:** a tabela criada assim existe apenas durante a execução desse Script. Ela não é automaticamente compartilhada com outros Scripts. Na próxima seção, vamos entender uma forma de compartilhar código e dados de configuração por meio de um ModuleScript.

## 7. Conhecendo o ModuleScript

Um `ModuleScript` permite organizar código que pode ser utilizado por outros Scripts. Ele não é executado automaticamente como um Script comum: outro Script precisa carregá-lo com `require()`.

Vamos criar um ModuleScript para centralizar valores iniciais do explorador.

Na pasta `Jogador`, adicione um `ModuleScript` chamado `ConfiguracaoJogador`.

````python
local ConfiguracaoJogador = {
    moedasIniciais = 10,
    chavesIniciais = 1,
    cristaisIniciais = 0,
    energiaInicial = 100
}

return ConfiguracaoJogador
````

O `return` disponibiliza a tabela para o Script que carregar o módulo.

## 8. Utilizando o ModuleScript com `require`

Agora, dentro da pasta `Jogador`, crie um `Script` chamado `TesteConfiguracao`.

O código abaixo carrega o ModuleScript e utiliza seus valores.

````python
local configuracao = require(script.Parent.ConfiguracaoJogador)

local jogador = {
    moedas = configuracao.moedasIniciais,
    chaves = configuracao.chavesIniciais,
    cristais = configuracao.cristaisIniciais,
    energia = configuracao.energiaInicial
}

print("JOGADOR CRIADO A PARTIR DO MODULO")
print("Moedas:", jogador.moedas)
print("Chaves:", jogador.chaves)
print("Cristais:", jogador.cristais)
print("Energia:", jogador.energia)
````

### Entendendo os caminhos

- `script` representa o Script que está executando;
- `script.Parent` representa a pasta que contém esse Script;
- `script.Parent.ConfiguracaoJogador` acessa o ModuleScript irmão;
- `require()` carrega o módulo e recebe o valor devolvido por `return`.

**Atenção:** os nomes e a hierarquia no Explorer precisam corresponder exatamente aos caminhos utilizados no código.

## 9. Separando as ações do jogador

Agora vamos criar um ModuleScript responsável pelas ações do explorador.

Na pasta `Jogador`, adicione um `ModuleScript` chamado `AcoesJogador`.

````python
local AcoesJogador = {}

function AcoesJogador.adicionarMoedas(jogador, quantidade)
    jogador.moedas = jogador.moedas + quantidade
end

function AcoesJogador.adicionarCristais(jogador, quantidade)
    jogador.cristais = jogador.cristais + quantidade
end

function AcoesJogador.mostrarStatus(jogador)
    print("====== STATUS DO JOGADOR ======")
    print("Moedas:", jogador.moedas)
    print("Cristais:", jogador.cristais)
    print("Energia:", jogador.energia)
end

return AcoesJogador
````

Neste módulo:
- criamos uma tabela vazia chamada `AcoesJogador`;
- adicionamos funções a essa tabela;
- cada função recebe a tabela `jogador` como parâmetro;
- no final, devolvemos a tabela com `return`.

Assim, outro Script poderá utilizar essas funções.

## 10. Chamando as ações de outro Script

Crie um `Script` chamado `TesteAcoes` dentro da pasta `Jogador`.

Ele carregará o módulo `AcoesJogador` e utilizará as funções.

````python
local AcoesJogador = require(script.Parent.AcoesJogador)

local jogador = {
    moedas = 10,
    cristais = 0,
    energia = 100
}

AcoesJogador.mostrarStatus(jogador)

AcoesJogador.adicionarMoedas(jogador, 15)
AcoesJogador.adicionarCristais(jogador, 2)

AcoesJogador.mostrarStatus(jogador)
````

Agora as ações estão separadas do Script que controla o teste. Isso facilita reutilizar e manter o código.

**Importante:** esse exemplo é didático e cria uma tabela local ao Script de teste. Cada Script que criar sua própria tabela terá seus próprios dados. Compartilhar o estado real de jogadores entre sistemas exige uma arquitetura adequada, que será desenvolvida em etapas futuras.

## 11. Organizando a missão em outra pasta

Na pasta `Missao`, crie um `Script` chamado `EtapasMissao`.

Ele será responsável por apresentar as etapas da aventura.

````python
local etapas = {
    "Floresta",
    "Rio",
    "Caverna",
    "Ponte",
    "Templo Antigo",
    "Portao do Tesouro"
}

print("ETAPAS DA MISSAO")

for i = 1, #etapas do
    print(i, "-", etapas[i])
end
````

Separar as etapas em outro Script deixa a organização mais clara: os arquivos da pasta `Jogador` cuidam dos dados e ações do explorador, enquanto os arquivos da pasta `Missao` cuidam da missão.

## 14. Cuidados ao executar os exemplos

Durante os testes, observe estas orientações:

1. Scripts comuns em `ServerScriptService` executam no servidor;
2. ModuleScripts precisam ser carregados com `require()`;
3. `script.Parent` representa o objeto que contém o Script atual;
4. o caminho utilizado em `require()` precisa corresponder à hierarquia do Explorer;
5. não coloque o mesmo teste em vários Scripts com nomes iguais sem necessidade;
6. para evitar mensagens duplicadas, mantenha apenas os Scripts de teste que estiver utilizando.

Os exemplos desta aula são independentes. Você pode testar um de cada vez, desativando ou removendo temporariamente os Scripts de teste que não estiverem sendo utilizados.
