# 💡 O que é uma Condição?

Uma condição é uma pergunta que o programa faz.

Exemplos:

- O jogador está vivo?
- O jogador tem energia?
- O jogador encontrou a chave?
- O jogador venceu a fase?

A resposta será:

✅ SIM

ou

❌ NÃO

---

# 🎮 Como os Jogos Utilizam Condições?

Jogos usam condições para:

- abrir portas
- liberar fases
- verificar vida
- controlar energia
- mostrar vitória
- mostrar derrota



# 🔀 Exemplo Visual

```plaintext
Tem energia?
      |
      +--- SIM ---> Pode correr
      |
      +--- NÃO ---> Sem energia
```

As condições funcionam exatamente dessa forma.



# 🟦 Estrutura IF

O IF executa uma ação quando a condição é verdadeira.


````

idade = 12

if idade >= 10:
    print("Você pode jogar")
````


# 🧠 Explicação

O programa pergunta:

A idade é maior ou igual a 10?

Se SIM:

```plaintext
Você pode jogar
```



# 💻 Exemplo 1 – Jogador Vivo


````

vida = 100

if vida > 0:
    print("Jogador Vivo")
````


# 💻 Exemplo 2 – Game Over


````

vida = 0

if vida <= 0:
    print("Game Over")
````


# 🟩 Estrutura ELSEIF

Permite criar várias possibilidades.


````

pontos = 80

if pontos >= 100:
    print("Ouro")

elif pontos >= 50:
    print("Prata")

else:
    print("Bronze")
````


# 🧠 Explicação

100 ou mais → Ouro

50 ou mais → Prata

Menor que 50 → Bronze


# 🎮 Aplicando no Roblox

Agora vamos utilizar condições dentro dos nossos jogos.



# 🟥 Bloco de Dano com Game Over


````

local vida = 30

script.Parent.Touched:Connect(function()

    vida = vida - 10

    if vida > 0 then
        print("Vida: " .. vida)
    else
        print("Game Over")
    end

end)
````


# 🟩 Bloco de Energia


````

local energia = 80

script.Parent.Touched:Connect(function()

    energia = energia + 20

    if energia >= 100 then
        print("Energia Máxima!")
    else
        print("Energia: " .. energia)
    end

end)
````


# 🟨 Sistema de Medalhas


````

local pontos = 0

script.Parent.Touched:Connect(function()

    pontos = pontos + 50

    if pontos >= 200 then
        print("🥇 Ouro")

    elseif pontos >= 100 then
        print("🥈 Prata")

    else
        print("🥉 Bronze")
    end

end)
````
