## 2. O que é uma função?

Uma função é um bloco de código com um nome. Ela só executa os comandos que estão dentro dela quando é chamada.

Observe a estrutura básica:

```lua
local function nomeDaFuncao()
    -- comandos da funcao
end
```

- `local function` inicia a definição de uma função local;
- `nomeDaFuncao` é o nome escolhido;
- `()` indicam onde podem ser informados parâmetros;
- `end` encerra a função.

Vamos criar nossa primeira função.

````python
local function mostrarMensagem()
    print("Bem-vindo a Ilha do Tesouro!")
end

mostrarMensagem()
````

### Entendendo o exemplo

Primeiro, definimos a função `mostrarMensagem`. Depois, usamos `mostrarMensagem()` para chamá-la.

**Atenção:** escrever a definição da função não significa executá-la. Para executar seu conteúdo, precisamos chamar a função.

## 3. Chamando uma função várias vezes

Uma das vantagens das funções é poder reutilizar o mesmo bloco de comandos.

````python
local function mostrarMensagem()
    print("========================")
    print("O explorador continua a aventura!")
    print("========================")
end

mostrarMensagem()
mostrarMensagem()
mostrarMensagem()
````

A mesma função foi chamada três vezes. Não precisamos copiar os três comandos `print` para cada execução: basta chamar a função novamente.

## 4. Funções com parâmetros

Um **parâmetro** permite que uma função receba uma informação. Assim, a função pode executar uma ação usando valores diferentes a cada chamada.

````python
local function cumprimentar(nome)
    print("Ola, " .. nome .. "!")
end

cumprimentar("Explorador")
cumprimentar("Guardiao")
cumprimentar("Aventureiro")
````

### O que está acontecendo?

- `nome` é o parâmetro definido na função;
- `"Explorador"` é o valor enviado na primeira chamada;
- `"Guardiao"` é o valor enviado na segunda chamada;
- `"Aventureiro"` é o valor enviado na terceira chamada.

O operador `..` junta textos em Lua. Essa operação é chamada de concatenação.

**Diferença importante:**
- **Parâmetro:** nome usado na definição da função;
- **Argumento:** valor enviado quando chamamos a função.

## 5. Funções com mais de um parâmetro

Uma função pode receber várias informações, separadas por vírgula.

````python
local function mostrarColeta(item, quantidade)
    print("Item encontrado:", item)
    print("Quantidade:", quantidade)
end

mostrarColeta("Moeda", 5)
mostrarColeta("Cristal", 2)
mostrarColeta("Chave", 1)
````

No exemplo, a função recebe dois parâmetros:
- `item`: o nome do recurso;
- `quantidade`: quantas unidades foram encontradas.

A cada chamada, enviamos argumentos diferentes.

## 6. Funções que alteram uma tabela

Vamos utilizar o inventário criado na Aplicação 5. A função `adicionarMoedas` receberá uma quantidade e atualizará a tabela.

````python
local inventario = {
    moedas = 10,
    chaves = 2,
    cristais = 1
}

local function adicionarMoedas(quantidade)
    inventario.moedas = inventario.moedas + quantidade
end

print("Moedas antes:", inventario.moedas)

adicionarMoedas(5)

print("Moedas depois:", inventario.moedas)
````

A função utiliza a tabela `inventario` criada fora dela. Quando chamamos `adicionarMoedas(5)`, o valor de moedas aumenta em 5.

A função evita repetir a operação de atualização em vários pontos do código.

## 7. Criando funções para diferentes recursos

Podemos criar uma função para adicionar moedas, outra para adicionar cristais e outra para utilizar uma chave.

````python
local inventario = {
    moedas = 10,
    chaves = 2,
    cristais = 1
}

local function adicionarMoedas(quantidade)
    inventario.moedas = inventario.moedas + quantidade
end

local function adicionarCristais(quantidade)
    inventario.cristais = inventario.cristais + quantidade
end

local function usarChave()
    inventario.chaves = inventario.chaves - 1
end

adicionarMoedas(10)
adicionarCristais(2)
usarChave()

print("Moedas:", inventario.moedas)
print("Cristais:", inventario.cristais)
print("Chaves:", inventario.chaves)
````

Observe como cada função tem uma responsabilidade:
- `adicionarMoedas()` altera as moedas;
- `adicionarCristais()` altera os cristais;
- `usarChave()` reduz a quantidade de chaves.

Por enquanto, vamos considerar que existe uma chave disponível quando chamamos `usarChave`. Em uma etapa futura, podemos aprender a validar essa condição.

## 8. A instrução `return`

Algumas funções precisam devolver um resultado. Para isso, usamos `return`.

Veja uma função que calcula o total de moedas depois de uma coleta.

````python
local function calcularMoedas(atual, encontradas)
    local total = atual + encontradas
    return total
end

local resultado = calcularMoedas(10, 5)

print("Total de moedas:", resultado)
````

### Entendendo o `return`

1. Chamamos `calcularMoedas(10, 5)`;
2. os valores chegam aos parâmetros `atual` e `encontradas`;
3. a função calcula a soma;
4. `return total` devolve o resultado;
5. o resultado é armazenado na variável `resultado`.

Uma função pode devolver um valor para ser usado em outra parte do programa.

## 9. Função que consulta o inventário

Vamos criar uma função que recebe o nome de um recurso e retorna a quantidade correspondente. Para manter o exemplo simples, vamos consultar os campos conhecidos da tabela.

````python
local inventario = {
    moedas = 25,
    chaves = 1,
    cristais = 3
}

local function consultarMoedas()
    return inventario.moedas
end

local function consultarChaves()
    return inventario.chaves
end

print("Moedas disponíveis:", consultarMoedas())
print("Chaves disponíveis:", consultarChaves())
````

As funções retornam os valores armazenados na tabela. Como elas usam `return`, podemos utilizar o resultado diretamente dentro de `print()`.

Mais adiante, poderemos estudar formas mais flexíveis de consultar diferentes campos com uma única função.

## 10. Aplicação no Roblox Studio — ações do explorador

Vamos criar um Script que representa três ações da aventura:
- receber moedas;
- encontrar cristais;
- utilizar uma chave.

**Passos:**
1. Abra o projeto Ilha do Tesouro no Roblox Studio;
2. no Explorer, localize `ServerScriptService`;
3. crie um `Script`;
4. renomeie para `AcoesExplorador`;
5. copie o código abaixo;
6. clique em **Play** e acompanhe o painel **Output**.

````python
local inventario = {
    moedas = 10,
    chaves = 2,
    cristais = 1
}

local function mostrarInventario()
    print("=== INVENTARIO DO EXPLORADOR ===")
    print("Moedas:", inventario.moedas)
    print("Chaves:", inventario.chaves)
    print("Cristais:", inventario.cristais)
end

local function coletarMoedas(quantidade)
    inventario.moedas = inventario.moedas + quantidade
    print("Moedas coletadas:", quantidade)
end

local function coletarCristais(quantidade)
    inventario.cristais = inventario.cristais + quantidade
    print("Cristais coletados:", quantidade)
end

local function usarChave()
    inventario.chaves = inventario.chaves - 1
    print("O explorador utilizou uma chave.")
end

mostrarInventario()

coletarMoedas(5)
coletarCristais(2)
usarChave()

mostrarInventario()
````

### Resultado esperado

O inventário começa com:
- 10 moedas;
- 2 chaves;
- 1 cristal.

Depois das ações, deverá apresentar:
- 15 moedas;
- 1 chave;
- 3 cristais.

Repare que `mostrarInventario()` foi chamada antes e depois das ações. A mesma função foi reutilizada para exibir os dados atualizados.
