# 🏝️ Aula 3 — Aplicação: A Trilha do Tesouro

## 🎮 Projeto Ilha do Tesouro — Aplicando o `for` no Roblox

Nesta aula vamos utilizar o **laço `for` dentro do nosso jogo**.

Na aula técnica aprendemos que o `for` permite repetir uma ação várias vezes.

Agora vamos transformar esse conhecimento em uma missão:

> 🗺️ **Criar uma trilha automática que conduz o explorador até o tesouro.**

### 🎯 Objetivos
- utilizar `for` em um projeto real;
- criar vários objetos automaticamente;
- utilizar o contador `i`;
- modificar a posição dos objetos;
- controlar a quantidade de objetos;
- criar uma trilha no cenário;
- personalizar os objetos.

### ⚠️ Nesta aula
Vamos trabalhar **somente com `for`**.

Ainda não vamos utilizar `while`, tabelas, funções ou sistemas complexos de pontuação.

# 🗺️ 1. A missão

Imagine que o explorador precisa encontrar um tesouro escondido.

Vamos construir uma trilha:

```text
🟫 → 🟫 → 🟫 → 🟫 → 🟫 → 🟫 → 🟫 → 🟫 → 🟫 → 🏆
```

Poderíamos criar cada bloco manualmente.

Mas hoje vamos ensinar o computador a fazer isso.

> "Crie 10 blocos e coloque cada um em uma posição diferente."

O `for` será responsável por repetir a criação.

# 🧱 2. Preparando o Roblox Studio

1. Abra o projeto da **Ilha do Tesouro**.
2. Vá até o `Workspace`.
3. Escolha uma área livre.
4. Crie um `Script`.
5. Coloque o Script dentro de `ServerScriptService`.

Vamos começar com um teste simples.

# 🧪 3. Primeiro teste — Criando 5 blocos

```lua
for i = 1, 5 do

    local bloco = Instance.new("Part")

    bloco.Size = Vector3.new(4, 1, 4)

    bloco.Position = Vector3.new(i * 5, 1, 0)

    bloco.Anchored = true

    bloco.Parent = workspace

end
```

Execute o jogo.

O Roblox criou cinco blocos.

O `for` repetiu:

```text
1 → cria bloco
2 → cria bloco
3 → cria bloco
4 → cria bloco
5 → cria bloco
```

# 🔍 4. Entendendo o contador `i`

Observe:

```lua
bloco.Position = Vector3.new(i * 5, 1, 0)
```

Quando `i = 1`, X vale `5`.

Quando `i = 2`, X vale `10`.

Quando `i = 3`, X vale `15`.

O contador está sendo utilizado para organizar os blocos.

> 💡 O `i` não serve apenas para contar. Ele também pode controlar posições e outras características do objeto.

# 🏝️ 5. Nossa primeira trilha

Agora vamos aumentar para 10 blocos:

```lua
for i = 1, 10 do

    local bloco = Instance.new("Part")

    bloco.Size = Vector3.new(3, 1, 3)

    bloco.Position = Vector3.new(i * 4, 1, 0)

    bloco.Anchored = true

    bloco.Parent = workspace

end
```

Execute e faça o personagem caminhar sobre a trilha.

# 🎨 6. Personalizando a trilha

Podemos alterar a aparência dos blocos:

```lua
bloco.Color = Color3.fromRGB(139, 69, 19)
bloco.Material = Enum.Material.Wood
```

Código completo:

```lua
for i = 1, 10 do

    local bloco = Instance.new("Part")

    bloco.Size = Vector3.new(3, 1, 3)
    bloco.Position = Vector3.new(i * 4, 1, 0)
    bloco.Anchored = true

    bloco.Color = Color3.fromRGB(139, 69, 19)
    bloco.Material = Enum.Material.Wood

    bloco.Parent = workspace

end
```

Agora temos uma verdadeira trilha de madeira.

# 🧪 7. Experimento — Quantidade de blocos

### 💡 Descoberta

# 🪙 9. Transformando blocos em moedas

Agora vamos aproximar a atividade do jogo.

Em vez de blocos, podemos criar objetos que representem moedas:

```lua
for i = 1, 10 do

    local moeda = Instance.new("Part")

    moeda.Shape = Enum.PartType.Cylinder
    moeda.Size = Vector3.new(1, 0.5, 1)
    moeda.Position = Vector3.new(i * 4, 2, 0)

    moeda.Anchored = true

    moeda.Parent = workspace

end
```

Neste momento não vamos fazer a moeda dar pontos.

O objetivo é entender como o `for` pode criar vários elementos do jogo.

# 🔍 10. O que o `for` está fazendo?

```text
i = 1  → cria moeda
i = 2  → cria moeda
i = 3  → cria moeda
i = 4  → cria moeda
...
i = 10 → cria moeda
```

O código dentro do `for` é executado novamente para cada valor de `i`.

> 🔁 **Uma instrução escrita uma vez pode ser executada várias vezes.**
