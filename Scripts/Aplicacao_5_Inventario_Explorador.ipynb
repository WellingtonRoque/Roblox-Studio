## 3. Criando a tabela de recursos

Primeiro, vamos criar uma tabela com a quantidade inicial de cada recurso do explorador.

```python
local inventario = {
    moedas = 10,
    chaves = 2,
    cristais = 1
}

print("INVENTARIO DO EXPLORADOR")
print("Moedas:", inventario.moedas)
print("Chaves:", inventario.chaves)
print("Cristais:", inventario.cristais)
```

### Entendendo

A variável `inventario` guarda uma tabela. Cada campo possui um nome e um valor:
- `moedas` começa com 10;
- `chaves` começa com 2;
- `cristais` começa com 1.

Executando o Script com **Play**, confira os valores no painel **Output**.

## 4. Criando a lista de objetos

Além das quantidades, o explorador pode carregar objetos específicos. Vamos criar uma lista com os itens que ele já encontrou.

```python
local itens = {
    "Mapa Antigo",
    "Tocha",
    "Bússola"
}

print("OBJETOS NA MOCHILA")
print(itens[1])
print(itens[2])
print(itens[3])
```

A lista `itens` utiliza posições numéricas:
- posição 1: Mapa Antigo;
- posição 2: Tocha;
- posição 3: Bússola.

Lembre-se: em Lua, a primeira posição é 1.

## 5. Mostrando todos os objetos com `for`

Em vez de escrever um `print` para cada posição, podemos percorrer a lista com o laço `for`.

```python
local itens = {
    "Mapa Antigo",
    "Tocha",
    "Bússola"
}

print("=== MOCHILA ===")

for i = 1, #itens do
    print(i, "-", itens[i])
end
```

O trecho `#itens` informa quantos objetos existem na lista. O laço percorre as posições de 1 até o tamanho da lista.

## 6. Encontrando um novo objeto

Durante a exploração, o personagem encontrou uma Poção de Energia. Vamos adicionar esse objeto ao final da lista com `table.insert()`.

```python
local itens = {
    "Mapa Antigo",
    "Tocha",
    "Bússola"
}

table.insert(itens, "Pocao de Energia")

print("OBJETOS ATUALIZADOS")

for i = 1, #itens do
    print(i, "-", itens[i])
end
```

Depois de executar, a lista terá quatro objetos. O novo item foi inserido na última posição.

**Experimente:** troque `"Pocao de Energia"` por `"Pedra Misteriosa"` e execute novamente.

## 7. Usando um objeto

O explorador utilizou a Tocha. Como ela estava na posição 2, vamos removê-la usando `table.remove()`.

```python
local itens = {
    "Mapa Antigo",
    "Tocha",
    "Bússola",
    "Pocao de Energia"
}

table.remove(itens, 2)

print("MOCHILA DEPOIS DE USAR A TOCHA")

for i = 1, #itens do
    print(i, "-", itens[i])
end
```

A Tocha foi removida da lista. Os objetos que estavam depois dela avançaram uma posição.

Atenção: `table.remove(itens, 2)` remove o objeto que estiver na posição 2. Se a lista mudar, o objeto nessa posição também poderá mudar.

## 8. Atualizando as quantidades de recursos

Agora o explorador encontrou um baú com moedas e cristais, mas precisou usar uma chave para abri-lo.

Vamos atualizar os valores da tabela `inventario`.

```python
local inventario = {
    moedas = 10,
    chaves = 2,
    cristais = 1
}

print("ANTES DO BAU")
print("Moedas:", inventario.moedas)
print("Chaves:", inventario.chaves)
print("Cristais:", inventario.cristais)

-- O explorador abriu o bau
inventario.chaves = inventario.chaves - 1

-- O bau tinha 15 moedas e 2 cristais
inventario.moedas = inventario.moedas + 15
inventario.cristais = inventario.cristais + 2

print("DEPOIS DO BAU")
print("Moedas:", inventario.moedas)
print("Chaves:", inventario.chaves)
print("Cristais:", inventario.cristais)
```

Resultado esperado:
- moedas: 25;
- chaves: 1;
- cristais: 3.

O código primeiro subtrai uma chave e depois adiciona os recursos encontrados no baú.
