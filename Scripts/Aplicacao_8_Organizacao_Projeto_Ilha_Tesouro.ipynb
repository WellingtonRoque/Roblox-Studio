## 1. O que vamos construir?

Na Aula 8, estudamos Script, LocalScript, ModuleScript, `require()` e `script.Parent`. Agora vamos aplicar esses conceitos ao projeto.

Vamos criar três áreas:
- **Jogador:** configuração e ações do explorador;
- **Missao:** etapas da aventura;
- **Mundo:** informações gerais da ilha.

Ao final, um Script principal carregará os módulos e coordenará uma demonstração.

Nesta aplicação, o foco é organização e reutilização de código. Não vamos implementar ainda coleta por toque, interfaces gráficas, salvamento ou comunicação entre cliente e servidor.

## 2. Preparando o Roblox Studio

1. Abra o projeto Ilha do Tesouro.
2. No Explorer, localize `ServerScriptService`.
3. Crie uma pasta chamada `Sistemas`.
4. Dentro dela, crie as pastas `Jogador`, `Missao` e `Mundo`.

Estrutura inicial:

```text
ServerScriptService
    Sistemas
        Jogador
        Missao
        Mundo
```

Todos os arquivos desta aplicação ficarão nessas pastas.

## 3. Configuração do explorador

Na pasta `Jogador`, crie um **ModuleScript** chamado `ConfiguracaoJogador`.

````python
local ConfiguracaoJogador = {
    nome = "Aventureiro",
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100,
    nivel = 1
}

return ConfiguracaoJogador
````

O módulo contém os dados iniciais do personagem. O `return` devolve a tabela para o Script que carregar esse ModuleScript.

## 4. Testando a configuração com `require()`

Na mesma pasta `Jogador`, crie um **Script** chamado `TesteConfiguracao`.

````python
local configuracao = require(script.Parent.ConfiguracaoJogador)

print("=== CONFIGURACAO DO EXPLORADOR ===")
print("Nome:", configuracao.nome)
print("Nivel:", configuracao.nivel)
print("Moedas:", configuracao.moedas)
print("Chaves:", configuracao.chaves)
print("Cristais:", configuracao.cristais)
print("Energia:", configuracao.energia)
````

`script.Parent` representa a pasta que contém o Script atual. Como `TesteConfiguracao` e `ConfiguracaoJogador` estão na mesma pasta, `script.Parent.ConfiguracaoJogador` encontra o módulo.

Clique em **Play** e confira o painel Output.

## 5. Criando o módulo de ações

Na pasta `Jogador`, crie um **ModuleScript** chamado `AcoesJogador`.

````python
local AcoesJogador = {}

function AcoesJogador.adicionarMoedas(jogador, quantidade)
    jogador.moedas = jogador.moedas + quantidade
    print("Moedas adicionadas:", quantidade)
end

function AcoesJogador.adicionarCristais(jogador, quantidade)
    jogador.cristais = jogador.cristais + quantidade
    print("Cristais adicionados:", quantidade)
end

function AcoesJogador.recuperarEnergia(jogador, quantidade)
    jogador.energia = jogador.energia + quantidade
    print("Energia recuperada:", quantidade)
end

function AcoesJogador.mostrarStatus(jogador)
    print("====== STATUS DO EXPLORADOR ======")
    print("Nome:", jogador.nome)
    print("Nivel:", jogador.nivel)
    print("Moedas:", jogador.moedas)
    print("Chaves:", jogador.chaves)
    print("Cristais:", jogador.cristais)
    print("Energia:", jogador.energia)
end

return AcoesJogador
````

O módulo reúne funções relacionadas às ações do explorador. Cada função recebe a tabela `jogador` como parâmetro. Assim, o módulo modifica os dados enviados pelo Script que o utiliza.

## 6. Testando as ações

Na pasta `Jogador`, crie um **Script** chamado `TesteAcoes`.

````python
local configuracao = require(script.Parent.ConfiguracaoJogador)
local AcoesJogador = require(script.Parent.AcoesJogador)

local jogador = {
    nome = configuracao.nome,
    nivel = configuracao.nivel,
    moedas = configuracao.moedas,
    chaves = configuracao.chaves,
    cristais = configuracao.cristais,
    energia = configuracao.energia
}

AcoesJogador.mostrarStatus(jogador)

print("=== AVENTURA ===")
AcoesJogador.adicionarMoedas(jogador, 15)
AcoesJogador.adicionarCristais(jogador, 2)
AcoesJogador.recuperarEnergia(jogador, 10)

print("=== STATUS ATUALIZADO ===")
AcoesJogador.mostrarStatus(jogador)
````

Resultado esperado após as ações: 25 moedas, 1 chave, 2 cristais e 110 de energia.

Observe que `TesteAcoes` criou uma tabela própria a partir da configuração. Cada Script que cria sua própria tabela terá seus próprios dados; o ModuleScript disponibiliza código e configuração, mas não cria automaticamente um estado compartilhado entre Scripts.

## 7. Organizando as etapas da missão

Na pasta `Missao`, crie um **ModuleScript** chamado `EtapasMissao`.

````python
local EtapasMissao = {
    "Floresta",
    "Rio",
    "Caverna",
    "Ponte",
    "Templo Antigo",
    "Portao do Tesouro"
}

return EtapasMissao
````

O módulo guarda a lista de etapas e a devolve para os Scripts que precisarem utilizá-la.

## 8. Testando as etapas

Na pasta `Missao`, crie um **Script** chamado `TesteMissao`.

````python
local etapas = require(script.Parent.EtapasMissao)

print("====== ETAPAS DA MISSAO ======")

for i = 1, #etapas do
    print("Etapa", i, "-", etapas[i])
end
````

O Script carrega a lista com `require()` e utiliza `for` para percorrer as posições da tabela.

## 9. Organizando as informações do mundo

Na pasta `Mundo`, crie um **ModuleScript** chamado `InformacoesMundo`.

````python
local InformacoesMundo = {
    nome = "Ilha do Tesouro",
    mensagemInicial = "Bem-vindo a ilha!",
    mensagemFinal = "Voce encontrou o tesouro!"
}

return InformacoesMundo
````

Agora, na pasta `Mundo`, crie um **Script** chamado `TesteMundo`.

````python
local mundo = require(script.Parent.InformacoesMundo)

print("====== MUNDO ======")
print("Nome:", mundo.nome)
print(mundo.mensagemInicial)
print(mundo.mensagemFinal)
````

As informações gerais do cenário ficam separadas dos dados do explorador e das etapas da missão.

## 10. Conferindo a estrutura do Explorer

Ao concluir as etapas anteriores, a estrutura deverá ficar semelhante a esta:

```text
ServerScriptService

            AcoesJogador (ModuleScript)
            TesteAcoes (Script)
        Missao
            EtapasMissao (ModuleScript)
            TesteMissao (Script)
        Mundo
            InformacoesMundo (ModuleScript)
            TesteMundo (Script)
```

Confira os nomes e a localização de cada arquivo. Os caminhos usados em `require()` precisam corresponder à hierarquia do Explorer.

## 11. Integração — Script principal de demonstração

Na pasta `Sistemas`, crie um **Script** chamado `DemonstracaoProjeto`.

Como ele está diretamente dentro de `Sistemas`, os caminhos abaixo acessam as pastas `Jogador`, `Missao` e `Mundo`.

````python
-- MODULOS DO JOGADOR
local pastaJogador = script.Parent.Jogador
local configuracao = require(pastaJogador.ConfiguracaoJogador)
local AcoesJogador = require(pastaJogador.AcoesJogador)

-- MODULO DA MISSAO
local pastaMissao = script.Parent.Missao
local etapas = require(pastaMissao.EtapasMissao)

-- MODULO DO MUNDO
local pastaMundo = script.Parent.Mundo
local mundo = require(pastaMundo.InformacoesMundo)

-- DADOS DESTA DEMONSTRACAO
local jogador = {
    nome = configuracao.nome,
    nivel = configuracao.nivel,
    moedas = configuracao.moedas,
    chaves = configuracao.chaves,
    cristais = configuracao.cristais,
    energia = configuracao.energia
}

print("================================")
print(mundo.nome)
print(mundo.mensagemInicial)
print("================================")

AcoesJogador.mostrarStatus(jogador)

print("====== INICIO DA MISSAO ======")

for i = 1, #etapas do
    print("Explorando:", etapas[i])

    if i == 2 then
        AcoesJogador.adicionarMoedas(jogador, 10)
    elseif i == 3 then
        AcoesJogador.adicionarCristais(jogador, 2)
    elseif i == 4 then
        AcoesJogador.recuperarEnergia(jogador, 15)
    end
end

print("====== FIM DA MISSAO ======")
AcoesJogador.mostrarStatus(jogador)
print(mundo.mensagemFinal)
````

### O que foi integrado?

O Script principal carrega os módulos responsáveis por cada parte:
- configuração e ações do jogador;
- etapas da missão;
- informações do mundo.

Ele coordena a demonstração, enquanto os módulos mantêm as responsabilidades separadas.
