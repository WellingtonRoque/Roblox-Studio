# 💡 O que é uma variável?

Uma variável é como uma caixa que guarda informações.

## Exemplos:

📦 nome → Wellington  
📦 pontos → 10  
📦 velocidade → 50  

---

# 🎮 Onde jogos usam variáveis?

Jogos utilizam variáveis o tempo todo:

- vida do personagem
- moedas
- pontuação
- velocidade
- energia
- inventário

---

# 💬 Perguntas para os alunos

- Como o jogo sabe quantas moedas o jogador pegou?
- Como o jogo sabe a vida do personagem?
- Como o jogo sabe a cor de um objeto?

👉 Tudo isso pode ser guardado em variáveis.



# 💻 Exemplo 1 – Guardando um Número


````

pontos = 10

print(pontos)
````


# 🧠 Explicação

- `pontos` → nome da variável
- `=` → recebe um valor
- `10` → valor guardado
- `print()` → mostra na tela



# 💻 Exemplo 2 – Guardando Texto


````

nome = "Alex"

print(nome)
````


# 🧠 Explicação

👉 Textos ficam entre aspas.



# 💻 Exemplo 3 – Fazendo Conta


````

vida = 100

print("Vida inicial:")
print(vida)

vida = vida - 20

print("Vida após dano:")
print(vida)
````


# 🧠 Explicação

👉 A variável pode mudar.

Isso acontece em jogos quando:
- o jogador perde vida
- ganha moedas
- recebe pontos



# 💻 Exemplo 4 – Sistema de Pontos


````

pontos = 0

print("Pontuação inicial:")
print(pontos)

pontos = pontos + 5

print("Pontuação atual:")
print(pontos)
````


# 🎮 Conectando com Roblox Studio

Agora vamos usar variáveis para alterar objetos dentro do jogo.



# 🟦 Exemplo Roblox 1 – Guardando uma Cor


````

local cor = Color3.new(1, 0, 0)

script.Parent.Color = cor
````


# 🧠 Explicação

- `local cor` → variável chamada cor
- `Color3.new()` → cria uma cor
- vermelho = `(1, 0, 0)`



# 🟦 Exemplo Roblox 2 – Guardando Velocidade


````

local velocidade = 50

print(velocidade)
````


# 🧠 Explicação

👉 Jogos usam variáveis para controlar velocidade.



# 🟦 Exemplo Roblox 3 – Mudando Cor ao Encostar


````

local cor = Color3.new(0, 1, 0)

script.Parent.Touched:Connect(function(hit)
    script.Parent.Color = cor
end)
````


# 🧠 O que o aluno aprende aqui

✔ variável pode guardar cor  
✔ variável pode ser reutilizada  
✔ variável ajuda a organizar o código
