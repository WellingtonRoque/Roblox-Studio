## 3. Criando os dados do explorador

Começaremos com uma tabela para organizar os recursos e a energia do personagem.

```python
local explorador = {
    moedas = 10,
    chaves = 2,
    cristais = 1,
    energia = 50
}

print("DADOS INICIAIS DO EXPLORADOR")
print("Moedas:", explorador.moedas)
print("Chaves:", explorador.chaves)
print("Cristais:", explorador.cristais)
print("Energia:", explorador.energia)
```

A tabela `explorador` guarda quatro informações. Vamos utilizá-la nas funções que criaremos a seguir.

## 4. Ação 1 — Mostrar o inventário

Vamos criar uma função responsável por apresentar os dados do explorador. Essa função poderá ser chamada sempre que quisermos conferir os recursos.

```python
local explorador = {
    moedas = 10,
    chaves = 2,
    cristais = 1,
    energia = 50
}

local function mostrarInventario()
    print("====== INVENTARIO ======")
    print("Moedas:", explorador.moedas)
    print("Chaves:", explorador.chaves)
    print("Cristais:", explorador.cristais)
    print("Energia:", explorador.energia)
end

mostrarInventario()
```

**Observe:** os comandos de exibição estão agrupados em `mostrarInventario()`. Quando chamamos essa função, todos os dados são apresentados.

## 5. Ação 2 — Coletar moedas

Agora vamos criar uma função que recebe a quantidade de moedas encontradas e atualiza o inventário.

```python
local explorador = {
    moedas = 10
}

local function coletarMoedas(quantidade)
    explorador.moedas = explorador.moedas + quantidade
    print("Moedas encontradas:", quantidade)
    print("Total de moedas:", explorador.moedas)
end

coletarMoedas(5)
coletarMoedas(3)
```

A função `coletarMoedas(quantidade)` recebe um parâmetro. Cada chamada envia um argumento diferente.

- Primeira chamada: adiciona 5 moedas;
- segunda chamada: adiciona mais 3 moedas.

O total passa de 10 para 18 moedas.

## 6. Ação 3 — Coletar cristais

Vamos criar outra função para adicionar cristais ao inventário.

```python
local explorador = {
    cristais = 1
}

local function coletarCristais(quantidade)
    explorador.cristais = explorador.cristais + quantidade
    print("Cristais encontrados:", quantidade)
    print("Total de cristais:", explorador.cristais)
end

coletarCristais(2)
coletarCristais(1)
```

A estrutura é semelhante à função de moedas. O que muda é o campo da tabela que será atualizado.

## 7. Ação 4 — Utilizar uma chave

O explorador precisa utilizar uma chave para abrir uma porta. Vamos criar uma função que diminui a quantidade de chaves.

Nesta primeira versão, a função considera que o explorador tem uma chave disponível.

```python
local explorador = {
    chaves = 2
}

local function usarChave()
    explorador.chaves = explorador.chaves - 1
    print("Uma chave foi utilizada.")
    print("Chaves restantes:", explorador.chaves)
end

usarChave()
```

A função `usarChave()` não precisa de parâmetros, porque a quantidade retirada é sempre uma unidade.

## 8. Ação 5 — Recuperar energia

Durante a aventura, o explorador pode encontrar uma fonte de energia. Vamos criar uma função que recebe a quantidade recuperada e atualiza o valor.

```python
local explorador = {
    energia = 40
}

local function recuperarEnergia(quantidade)
    explorador.energia = explorador.energia + quantidade
    print("Energia recuperada:", quantidade)
    print("Energia atual:", explorador.energia)
end

recuperarEnergia(10)
recuperarEnergia(5)
```

A energia começa em 40. Depois das duas recuperações, chega a 55.

Nesta atividade, não vamos limitar a energia a um valor máximo. O foco é praticar parâmetros e atualização de tabelas.

## 9. Ação 6 — Calcular moedas com bônus

Algumas missões podem oferecer um bônus. Vamos criar uma função que recebe a quantidade atual e o bônus e devolve o resultado usando `return`.

```python
local function calcularMoedasComBonus(moedas, bonus)
    local total = moedas + bonus
    return total
end

local resultado = calcularMoedasComBonus(20, 10)

print("Moedas com bonus:", resultado)
```

A função calcula e devolve o resultado. A variável `resultado` recebe o valor retornado.

O `return` é útil quando queremos que uma função produza um valor para ser utilizado em outro comando.

## 10. Integrando todas as ações

Agora vamos reunir as funções em um único Script. Este será o principal exercício prático da aplicação.

Substitua o conteúdo do Script `AcoesDoExplorador` pelo código completo abaixo.

```python
-- DADOS DO EXPLORADOR
local explorador = {
    moedas = 10,
    chaves = 2,
    cristais = 1,
    energia = 50
}

-- MOSTRAR INVENTARIO E ENERGIA
local function mostrarInventario()
    print("====== STATUS DO EXPLORADOR ======")
    print("Moedas:", explorador.moedas)
    print("Chaves:", explorador.chaves)
    print("Cristais:", explorador.cristais)
    print("Energia:", explorador.energia)
end

-- COLETAR MOEDAS
local function coletarMoedas(quantidade)
    explorador.moedas = explorador.moedas + quantidade
    print("O explorador encontrou", quantidade, "moedas.")
end

-- COLETAR CRISTAIS
local function coletarCristais(quantidade)
    explorador.cristais = explorador.cristais + quantidade
    print("O explorador encontrou", quantidade, "cristais.")
end

-- UTILIZAR CHAVE
local function usarChave()
    explorador.chaves = explorador.chaves - 1
    print("O explorador abriu uma porta.")
end

-- RECUPERAR ENERGIA
local function recuperarEnergia(quantidade)
    explorador.energia = explorador.energia + quantidade
    print("O explorador recuperou", quantidade, "de energia.")
end

-- CALCULAR BONUS
local function calcularMoedasComBonus(moedas, bonus)
    local total = moedas + bonus
    return total
end

-- INICIO DA AVENTURA
print("===== INICIO DA AVENTURA =====")
mostrarInventario()

print("===== EVENTO 1: BAU DE TESOURO =====")
coletarMoedas(15)
coletarCristais(2)

print("===== EVENTO 2: PORTA ANTIGA =====")
usarChave()

print("===== EVENTO 3: FONTE DE ENERGIA =====")
recuperarEnergia(20)

print("===== EVENTO 4: RECOMPENSA =====")
local totalComBonus = calcularMoedasComBonus(explorador.moedas, 5)
print("Moedas incluindo bonus:", totalComBonus)

print("===== STATUS FINAL =====")
mostrarInventario()
```

### Resultado esperado

O explorador começa com:
- 10 moedas;
- 2 chaves;
- 1 cristal;
- 50 de energia.

Durante a aventura:
- encontra 15 moedas;
- encontra 2 cristais;
- utiliza uma chave;
- recupera 20 de energia;
- calcula um bônus de 5 moedas.

Ao final, o inventário deverá apresentar:
- **25 moedas**;
- **1 chave**;
- **3 cristais**;
- **70 de energia**.

O cálculo com bônus apresenta 30 moedas, mas não altera o inventário: a função apenas calcula e retorna o resultado.
