## 3. Etapa 1 — Criando o personagem e seus recursos

Começaremos com uma tabela para guardar as informações do explorador.

```python
local explorador = {
    nome = "Aventureiro",
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100,
    nivel = 1
}

print("PERSONAGEM:", explorador.nome)
print("Nivel:", explorador.nivel)
print("Moedas:", explorador.moedas)
print("Chaves:", explorador.chaves)
print("Cristais:", explorador.cristais)
print("Energia:", explorador.energia)
```

Incluímos o campo `nivel`, que será atualizado quando a missão for concluída. A tabela mantém os dados relacionados ao personagem em um único lugar.

## 4. Etapa 2 — Criando as etapas da missão

A missão terá seis locais. Vamos guardar os nomes em uma lista e percorrer essa lista com `for`.

```python
local etapas = {
    "Floresta",
    "Rio",
    "Caverna",
    "Ponte",
    "Templo Antigo",
    "Portao do Tesouro"
}

for i = 1, #etapas do
    print("Etapa", i, "-", etapas[i])
end
```

A lista `etapas` organiza a sequência da aventura. O laço `for` percorre todas as posições, e `#etapas` informa o número de elementos.

## 5. Etapa 3 — Criando funções de apoio

Vamos criar funções para mostrar o status e coletar recursos. Essas funções serão reutilizadas durante a missão.

```python
local explorador = {
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100
}

local function mostrarStatus()
    print("====== STATUS DO EXPLORADOR ======")
    print("Moedas:", explorador.moedas)
    print("Chaves:", explorador.chaves)
    print("Cristais:", explorador.cristais)
    print("Energia:", explorador.energia)
end

local function coletarMoedas(quantidade)
    explorador.moedas = explorador.moedas + quantidade
    print("Moedas coletadas:", quantidade)
end

local function coletarCristais(quantidade)
    explorador.cristais = explorador.cristais + quantidade
    print("Cristais coletados:", quantidade)
end

mostrarStatus()
coletarMoedas(5)
coletarCristais(1)
mostrarStatus()
```

As funções `coletarMoedas()` e `coletarCristais()` recebem a quantidade como parâmetro e atualizam os campos da tabela.

## 6. Etapa 4 — Consumindo energia durante a exploração

Cada etapa exige esforço. Vamos criar uma função que diminui a energia e informa quanto restou.

```python
local explorador = {
    energia = 100
}

local function consumirEnergia(quantidade)
    explorador.energia = explorador.energia - quantidade
    print("Energia consumida:", quantidade)
    print("Energia restante:", explorador.energia)
end

consumirEnergia(10)
consumirEnergia(15)
```

Nesta versão, a função apenas realiza a subtração. Na integração final, vamos verificar se ainda há energia antes de continuar a missão.

## 7. Etapa 5 — Verificando se o explorador pode continuar

Agora vamos combinar função, condicional e `return`. A função verifica se o explorador possui energia maior que zero.

```python
local explorador = {
    energia = 20
}

local function podeContinuar()
    if explorador.energia > 0 then
        return true
    else
        return false
    end
end

if podeContinuar() then
    print("O explorador pode continuar.")
else
    print("O explorador precisa recuperar energia.")
end
```

A função retorna `true` ou `false`. O `if` utiliza esse resultado para decidir qual mensagem mostrar.

## 8. Etapa 6 — Abrindo o portão do tesouro

O portão exige uma chave. A função deverá verificar a quantidade disponível, consumir uma chave se houver e retornar se conseguiu abrir.

```python
local explorador = {
    chaves = 1
}

local function abrirPortao()
    if explorador.chaves > 0 then
        explorador.chaves = explorador.chaves - 1
        print("Portao aberto!")
        return true
    else
        print("Voce nao possui uma chave.")
        return false
    end
end

local portaoAberto = abrirPortao()

if portaoAberto then
    print("Acesso liberado ao tesouro.")
else
    print("A missao continua bloqueada.")
end
```

A função não apenas mostra uma mensagem: ela também altera o inventário e devolve um resultado que pode ser usado em outra decisão.

## 9. Etapa 7 — Definindo eventos em cada local

Vamos utilizar uma tabela para indicar o que acontece em algumas etapas. Para manter a atividade no nível atual, os eventos serão definidos por condições com o contador do laço.

```python
for etapa = 1, 6 do
    print("Explorando etapa:", etapa)

    if etapa == 2 then
        print("Evento: encontrou moedas!")
    elseif etapa == 3 then
        print("Evento: encontrou cristais!")
    elseif etapa == 4 then
        print("Evento: encontrou uma fonte de energia!")
    elseif etapa == 6 then
        print("Evento: chegou ao portao do tesouro!")
    else
        print("Evento: exploracao da regiao.")
    end
end
```

A variável `etapa` informa em qual posição da missão estamos. As condições determinam qual evento será apresentado.

## 10. Projeto integrado — Sistema de Progressão

Agora vamos unir as partes em um único Script. Este será o principal resultado da aplicação.

**Antes de executar:** substitua o conteúdo do Script `SistemaDeProgressao` pelo código completo abaixo.

```python
-- =========================================
-- DADOS DO EXPLORADOR
-- =========================================

local explorador = {
    nome = "Aventureiro",
    moedas = 10,
    chaves = 1,
    cristais = 0,
    energia = 100,
    nivel = 1
}

-- =========================================
-- ETAPAS DA MISSAO
-- =========================================

local etapas = {
    "Floresta",
    "Rio",
    "Caverna",
    "Ponte",
    "Templo Antigo",
    "Portao do Tesouro"
}

-- =========================================
-- FUNCOES DO SISTEMA
-- =========================================

local function mostrarStatus()
    print("========== STATUS ==========")
    print("Explorador:", explorador.nome)
    print("Nivel:", explorador.nivel)
    print("Moedas:", explorador.moedas)
    print("Chaves:", explorador.chaves)
    print("Cristais:", explorador.cristais)
    print("Energia:", explorador.energia)
end

local function coletarMoedas(quantidade)
    explorador.moedas = explorador.moedas + quantidade
    print("Coletou", quantidade, "moedas.")
end

local function coletarCristais(quantidade)
    explorador.cristais = explorador.cristais + quantidade
    print("Coletou", quantidade, "cristais.")
end

local function consumirEnergia(quantidade)
    explorador.energia = explorador.energia - quantidade
    print("Energia restante:", explorador.energia)
end

local function podeContinuar()
    if explorador.energia > 0 then
        return true
    else
        return false
    end
end

local function abrirPortao()
    if explorador.chaves > 0 then
        explorador.chaves = explorador.chaves - 1
        print("Portao aberto!")
        return true
    else
        print("Falta uma chave para abrir o portao.")
        return false
    end
end

-- =========================================
-- INICIO DA MISSAO
-- =========================================

print("================================")
print("      ILHA DO TESOURO")
print("================================")

mostrarStatus()

local missaoConcluida = false

-- Percorre as etapas da aventura
for i = 1, #etapas do

    if podeContinuar() then
        print("--------------------------------")
        print("Etapa", i, "-", etapas[i])

        consumirEnergia(10)

        -- Eventos em locais específicos
        if i == 2 then
            coletarMoedas(15)

        elseif i == 3 then
            coletarCristais(2)

        elseif i == 4 then
            explorador.energia = explorador.energia + 20
            print("Encontrou uma fonte: +20 de energia.")

        elseif i == 5 then
            coletarMoedas(10)
            coletarCristais(1)

        elseif i == 6 then
            print("Chegou ao portao final.")
            missaoConcluida = abrirPortao()
        end

    else
        print("Energia insuficiente. A exploracao foi interrompida.")
        break
    end
end

-- =========================================
-- RECOMPENSA E PROGRESSAO
-- =========================================

if missaoConcluida then
    print("================================")
    print("       MISSAO CONCLUIDA!")
    print("================================")

    explorador.nivel = explorador.nivel + 1
    explorador.moedas = explorador.moedas + 20

    print("Subiu para o nivel:", explorador.nivel)
    print("Recebeu 20 moedas de recompensa.")
else
    print("================================")
    print("       MISSAO NAO CONCLUIDA")
    print("================================")
end

-- =========================================
-- STATUS FINAL
-- =========================================

mostrarStatus()
print("===== FIM DA AVENTURA =====")
```

### Resultado esperado

Com os valores iniciais do exemplo, o explorador deverá:

- começar no nível 1, com 10 moedas, 1 chave e 100 de energia;
- percorrer as seis etapas;
- encontrar 25 moedas e 3 cristais;
- recuperar 20 pontos de energia na fonte;
- abrir o portão usando sua chave;
- concluir a missão;
- avançar para o nível 2;
- receber 20 moedas de recompensa.

**Status final esperado:**
- nível: 2;
- moedas: 65;
- chaves: 0;
- cristais: 3;
- energia: 50.

Confira os valores no Output. Se alterar os eventos ou os valores iniciais, o resultado também mudará.
