# 1. 🤔 Por que precisamos repetir?

Imagine que queremos mostrar cinco mensagens:

```python
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
```

Funciona. Mas e se tivermos **100 moedas**?

Precisaríamos escrever a mesma instrução muitas vezes.

É para situações como essa que utilizamos estruturas de repetição.

````python
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
print("Moeda encontrada!")
````

# 2. 🔁 Nossa primeira repetição

```python
for i in range(5):
    print("Moeda encontrada!")
```

- `for` inicia a repetição.
- `i` é o contador.
- `range(5)` determina a quantidade de posições.
- `:` inicia o bloco.
- O código indentado será repetido.

````python
for i in range(5):
    print("Moeda encontrada!")
````

# 3. 🔢 Conhecendo o contador

```python
for i in range(5):
    print(i)
```

O resultado é:

```text
0
1
2
3
4
```

**Atenção:** `range(5)` não inclui o número 5.

````python
for i in range(5):
    print(i)
````

# 4. 🪙 Contando as moedas

```python
for i in range(1, 11):
    print("Moeda", i)
```

`range(1, 11)` começa em 1 e vai até antes do 11.

````python
for i in range(1, 11):
    print("Moeda", i)
````

# 7. ⏪ Contagem regressiva

Também podemos fazer o contador diminuir.

```python

```

O `-1` significa diminuir 1 a cada repetição.

````python
for i in range(10, 0, -1):
    print(i)
````

# 10. 🎮 O `for` no Roblox

Em Lua, a estrutura é parecida:

```lua
for i = 1, 5 do
    print("Moeda encontrada!")
end
```

### Python
```python
for i in range(1, 6):
    print("Moeda")
```

### Lua
```lua
for i = 1, 5 do
    print("Moeda")
end
```

No Roblox:
- `for` inicia a repetição;
- `i` é o contador;
- `do` inicia o bloco;
- `end` encerra o bloco.
