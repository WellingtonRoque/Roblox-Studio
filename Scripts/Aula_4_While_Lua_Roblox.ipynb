# 1. Por que precisamos do `while`?

Na aula anterior usamos `for` quando sabíamos quantas vezes repetir. Agora imagine: queremos continuar mostrando uma mensagem enquanto o explorador ainda tiver energia. A repetição depende de uma condição, por isso usamos `while`.

# 2. Estrutura do `while`

```lua
while condicao do
    -- comandos repetidos
end
```

- `while`: inicia a repetição.
- `condicao`: decide se continua.
- `do`: inicia o bloco.
- `end`: encerra o bloco.

Se a condição for verdadeira (`true`), executa. Se for falsa (`false`), sai do laço.

# 3. Primeiro exemplo: contar até 5

O contador começa em 1. A cada repetição, mostramos seu valor e aumentamos 1. Quando chega a 6, a condição `contador <= 5` fica falsa e o laço termina.

````lua
local contador = 1

while contador <= 5 do
    print(contador)

    contador = contador + 1
end
````

# 5. Moedas da Ilha do Tesouro

O laço continua enquanto a quantidade de moedas for menor que 5. Quando chegar a 5, a condição fica falsa.

````lua
local moedas = 0

while moedas < 5 do
    moedas = moedas + 1
    print("Moedas encontradas:", moedas)
end

print("Missão concluída!")
````

# 6. Energia do explorador

Agora a variável diminui. A repetição continua enquanto a energia for maior que zero.

````lua
local energia = 100

while energia > 0 do
    print("Energia:", energia)
    energia = energia - 20
end

print("O explorador ficou sem energia!")
````

# 7. O `while` com tempo

No Roblox, `task.wait(1)` faz o script aguardar aproximadamente 1 segundo. O número é expresso em segundos.

````lua
local contador = 1

while contador <= 5 do
    print("Aguardando...")
    task.wait(1)
    contador = contador + 1
end

print("Tempo encerrado!")
````

# 9. Repetição contínua: `while true`

`true` é sempre verdadeiro. Por isso, o laço continua indefinidamente. Esse formato pode ser usado em comportamentos contínuos, como luzes piscando e cristais brilhando.

**Atenção:** utilize `task.wait()` para não manter o script repetindo sem pausa.

````lua
while true do
    print("A Ilha está viva!")
    task.wait(2)
end
````

# 13. Comparando `for` e `while`

| `for` | `while` |
|---|---|
| Usamos quando sabemos quantas repetições queremos. | Usamos quando a repetição depende de uma condição. |
| Exemplo: criar 10 blocos. | Exemplo: continuar enquanto houver energia. |

**FOR:** quantidade determinada.  
**WHILE:** condição controla a continuidade.
